```python

import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load the dataset
iris = load_iris()
X = iris.data
y = iris.target

# Keep a final test set separate from model selection
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Number of features:", X_train.shape[1])
print("Classes:", list(iris.target_names))
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

# Scaling and classification are combined into pipelines
knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(n_neighbors=3))
])

svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", SVC(kernel="linear"))
])

print("Scaled KNN and SVM pipelines are ready.")
```

```python

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.tree import DecisionTreeClassifier

# Five folds, preserving class proportions
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models = {
    "Scaled KNN": knn_pipeline,
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Scaled SVM": svm_pipeline
}

scoring = {
    "accuracy": "accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro"
}

cv_results = []

for name, estimator in models.items():
    scores = cross_validate(
        estimator,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring
    )

    cv_results.append({
        "Algorithm": name,
        "Mean Accuracy (%)": scores["test_accuracy"].mean() * 100,
        "Accuracy Std (%)": scores["test_accuracy"].std() * 100,
        "Mean Precision (%)": scores["test_precision_macro"].mean() * 100,
        "Mean Recall (%)": scores["test_recall_macro"].mean() * 100,
        "Mean F1 (%)": scores["test_f1_macro"].mean() * 100
    })

cv_df = pd.DataFrame(cv_results)

print(cv_df.round(2).to_string(index=False))
```

```python

best_name = cv_df.loc[
    cv_df["Mean Accuracy (%)"].idxmax(),
    "Algorithm"
]

best_model = models[best_name]

print("Selected model:", best_name)
print("Selection criterion: Highest mean CV accuracy")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Fit only on the training data
best_model.fit(X_train, y_train)

# Final test predictions
y_test_pred = best_model.predict(X_test)

print("Selected model:", best_name)
print(
    "Final test accuracy:",
    round(accuracy_score(y_test, y_test_pred) * 100, 2),
    "%"
)

print("\nClassification report:")
print(classification_report(
    y_test,
    y_test_pred,
    target_names=iris.target_names
))

print("Confusion matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    cv_df["Algorithm"],
    cv_df["Mean Accuracy (%)"],
    yerr=cv_df["Accuracy Std (%)"],
    capsize=5
)

plt.ylabel("Cross-validation accuracy (%)")
plt.xlabel("Classification algorithm")
plt.title("5-Fold Cross-Validation Model Comparison")
plt.ylim(0, 105)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Display cross-validation results
print("CROSS-VALIDATION RESULTS")
display(cv_df.round(2))

# Select the model with the highest mean CV accuracy
best_name = cv_df.loc[
    cv_df["Mean Accuracy (%)"].idxmax(),
    "Algorithm"
]

best_model = models[best_name]

print("\nSelected model:", best_name)

# Fit on the complete training set
best_model.fit(X_train, y_train)

# Evaluate once on the held-out test set
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_test_pred = best_model.predict(X_test)

test_accuracy = accuracy_score(y_test, y_test_pred)

print(f"\nFinal test accuracy: {test_accuracy * 100:.2f}%")

print("\nClassification report:")
print(classification_report(
    y_test,
    y_test_pred,
    target_names=iris.target_names
))

print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

```python

# Save model comparison and final evaluation
cv_df.round(2).to_csv(
    "cross_validation_results.csv",
    index=False
)

final_results = pd.DataFrame({
    "Metric": [
        "Selected algorithm",
        "Cross-validation mean accuracy (%)",
        "Final test accuracy (%)"
    ],
    "Value": [
        best_name,
        round(
            cv_df.loc[
                cv_df["Algorithm"] == best_name,
                "Mean Accuracy (%)"
            ].iloc[0],
            2
        ),
        round(test_accuracy * 100, 2)
    ]
})

final_results.to_csv(
    "final_model_evaluation.csv",
    index=False
)

print("Evaluation files saved successfully!")
display(final_results)
```

```python

import gradio as gr
import numpy as np

# Iris class names
species_names = {
    0: "Iris setosa",
    1: "Iris versicolor",
    2: "Iris virginica"
}

species_descriptions = {
    "Iris setosa": "Typically has short petals and is clearly distinct from the other two species.",
    "Iris versicolor": "Typically has medium-sized petals and intermediate measurements.",
    "Iris virginica": "Typically has longer and wider petals than the other two species."
}

def predict_flower(sepal_length, sepal_width, petal_length, petal_width):
    values = [sepal_length, sepal_width, petal_length, petal_width]

    if any(v is None for v in values):
        return "### Please enter all four measurements."

    if any(v <= 0 for v in values):
        return "### Measurements must be greater than zero."

    sample = np.array([values], dtype=float)

    # Use the trained model or preprocessing pipeline
    prediction = int(app_model.predict(sample)[0])
    species = species_names.get(prediction, str(prediction))
    description = species_descriptions.get(species, "")

    return f"""
## 🌸 Prediction: {species}

{description}

---
**Input measurements (cm)**

- Sepal length: {sepal_length}
- Sepal width: {sepal_width}
- Petal length: {petal_length}
- Petal width: {petal_width}

*Prediction generated by your trained machine-learning model.*
"""

custom_css = """
.gradio-container {
    max-width: 1000px !important;
    margin: auto !important;
}
"""

with gr.Blocks(
    theme=gr.themes.Soft(),
    css=custom_css,
    title="Iris Species Classifier"
) as demo:

    gr.Markdown("""
    # 🌸 Iris Species Classifier
    ### AI Project 2 — Data Classification Using AI

    Enter the flower's four measurements in centimetres.
    The trained machine-learning model will predict its Iris species.
    """)

    with gr.Row():
        with gr.Column():
            gr.Markdown("### 📏 Flower measurements")

            sepal_length = gr.Slider(
                minimum=4.0, maximum=8.0, value=5.1,
                step=0.1, label="Sepal length (cm)"
            )
            sepal_width = gr.Slider(
                minimum=1.5, maximum=4.5, value=3.5,
                step=0.1, label="Sepal width (cm)"
            )
            petal_length = gr.Slider(
                minimum=1.0, maximum=7.0, value=1.4,
                step=0.1, label="Petal length (cm)"
            )
            petal_width = gr.Slider(
                minimum=0.1, maximum=2.5, value=0.2,
                step=0.1, label="Petal width (cm)"
            )

            predict_btn = gr.Button(
                "🔍 Predict Iris Species",
                variant="primary"
            )
            clear_btn = gr.Button("Reset measurements")

        with gr.Column():
            gr.Markdown("### 🧠 Model prediction")
            result = gr.Markdown(
                "Your prediction will appear here after you click **Predict Iris Species**."
            )

    gr.Markdown("### 🧪 Try sample flowers")

    gr.Examples(
        examples=[
            [5.1, 3.5, 1.4, 0.2],
            [6.0, 2.9, 4.5, 1.5],
            [6.5, 3.0, 5.5, 1.8]
        ],
        inputs=[
            sepal_length, sepal_width,
            petal_length, petal_width
        ],
        label="Select an example to load its measurements"
    )

    predict_btn.click(
        fn=predict_flower,
        inputs=[
            sepal_length, sepal_width,
            petal_length, petal_width
        ],
        outputs=result
    )

    clear_btn.click(
        fn=lambda: (5.1, 3.5, 1.4, 0.2,
                    "Your prediction will appear here after you click **Predict Iris Species**."),
        inputs=[],
        outputs=[
            sepal_length, sepal_width,
            petal_length, petal_width, result
        ]
    )

    gr.Markdown("""
    ---
    **Project workflow:** Iris dataset → Data preprocessing → Model training
    → Model evaluation → Species prediction

    *Educational demonstration. Predictions depend on the training data and model performance.*
    """)

demo.launch(share=True)
```
